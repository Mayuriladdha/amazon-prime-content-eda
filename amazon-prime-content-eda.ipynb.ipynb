## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# Write your code to make your dataset analysis ready.
# ==========================
# Data Wrangling Code
# ==========================

# Create copies of the original datasets
titles = titles_df.copy()
credits = credits_df.copy()

# Check dataset shapes
print("Titles Shape:", titles.shape)
print("Credits Shape:", credits.shape)

# Check duplicate values
print("\nDuplicate Rows in Titles:", titles.duplicated().sum())
print("Duplicate Rows in Credits:", credits.duplicated().sum())

# Remove duplicate rows
titles.drop_duplicates(inplace=True)
credits.drop_duplicates(inplace=True)

# Check missing values
print("\nMissing Values in Titles Dataset:")
print(titles.isnull().sum())

print("\nMissing Values in Credits Dataset:")
print(credits.isnull().sum())

# Convert release_year to integer (if required)
titles['release_year'] = titles['release_year'].astype('Int64')

# Convert runtime to numeric
titles['runtime'] = pd.to_numeric(titles['runtime'], errors='coerce')

# Fill missing values
titles['age_certification'].fillna('Not Rated', inplace=True)
titles['description'].fillna('No Description', inplace=True)
titles['seasons'].fillna(0, inplace=True)

credits['character'].fillna('Unknown', inplace=True)

# Verify data types
print("\nTitles Data Types:")
print(titles.dtypes)

print("\nCredits Data Types:")
print(credits.dtypes)

# Final dataset shape
print("\nFinal Titles Shape:", titles.shape)
print("Final Credits Shape:", credits.shape)
````

Data types were verified and corrected, with columns such as release_year and runtime converted into appropriate numeric formats for accurate analysis.

````python
# Chart 3: Top 10 Most Popular Genres

from collections import Counter

# Split genres and count their frequency
genre_list = titles_df['genres'].dropna().astype(str)

all_genres = []
for genres in genre_list:
    genres = genres.strip("[]").replace("'", "").split(", ")
    all_genres.extend(genres)

genre_count = Counter(all_genres)

genre_df = pd.DataFrame(
    genre_count.items(),
    columns=['Genre', 'Count']
).sort_values(by='Count', ascending=False).head(10)

# Plot
plt.figure(figsize=(12,6))
sns.barplot(data=genre_df, x='Count', y='Genre', palette='viridis')

plt.title('Top 10 Most Popular Genres on Amazon Prime', fontsize=16)
plt.xlabel('Number of Titles')
plt.ylabel('Genre')

plt.show()
````

````python
# Chart - 5 visualization code
# Chart 5: Top 10 Production Countries

from collections import Counter

country_list = titles_df['production_countries'].dropna().astype(str)

all_countries = []

for countries in country_list:
    countries = countries.strip("[]").replace("'", "").split(", ")
    all_countries.extend(countries)

country_count = Counter(all_countries)

country_df = pd.DataFrame(
    country_count.items(),
    columns=['Country', 'Count']
).sort_values(by='Count', ascending=False).head(10)

# Plot
plt.figure(figsize=(12,6))

sns.barplot(
    data=country_df,
    x='Count',
    y='Country',
    palette='magma'
)

plt.title('Top 10 Production Countries on Amazon Prime', fontsize=16)
plt.xlabel('Number of Titles')
plt.ylabel('Production Country')

plt.show()
````

#### Chart - 14 - Correlation Heatmap

````python
# Correlation Heatmap visualization code


plt.figure(figsize=(10,8))

# Select numerical columns
numeric_df = titles_df.select_dtypes(include=['int64', 'float64'])

# Correlation matrix
corr_matrix = numeric_df.corr()

# Plot heatmap
sns.heatmap(
    corr_matrix,
    annot=True,
    cmap='coolwarm',
    fmt='.2f',
    linewidths=0.5
)

plt.title('Correlation Heatmap of Numerical Variables', fontsize=16)

plt.show()
````

##### 1. Why did you pick the specific chart?

A Correlation Heatmap was selected because it is one of the most effective visualizations for understanding the relationships between multiple numerical variables simultaneously. Unlike individual scatter plots that compare only two variables at a time, a heatmap summarizes the correlation between every pair of numerical features in a single figure. It uses color intensity to indicate the strength and direction of the relationship, making it easy to identify strong positive, strong negative, and weak correlations. This visualization is particularly useful during Exploratory Data Analysis (EDA) because it helps detect patterns, identify redundant variables, and discover important relationships that may influence business decisions.

##### 2. What is/are the insight(s) found from the chart?

The correlation heatmap provides several valuable insights into the numerical variables present in the Amazon Prime dataset.

Key insights include:

- Variables with correlation values close to +1 have a strong positive relationship, meaning they tend to increase together.

- Variables with correlation values close to -1 have a strong negative relationship, indicating that as one variable increases, the other decreases.

- Correlation values near 0 suggest little or no linear relationship between the variables.

- IMDb Score and TMDb Score often show a moderate to strong positive correlation, indicating that highly rated titles on one platform are generally rated well on the other.

- IMDb Votes may show a positive relationship with TMDb Popularity, suggesting that widely viewed titles tend to receive more audience attention across platforms.

- Runtime, release year, and popularity generally exhibit weaker correlations, indicating that these variables influence content performance independently.
